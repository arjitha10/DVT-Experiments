```python
import pandas as pd
df = pd.read_csv(
    "/kaggle/input/datasets/imakash3011/customer-personality-analysis/marketing_campaign.csv",
    sep="\t")
print(df.head())
```

```python
print(df.shape)
```

```python
print(df.dtypes)
```

```python
high_income = df[df["Income"] > 50000]

print(high_income)
```

```python
graduates = df[df["Education"] == "Graduation"]

print(graduates)
```

```python
sorted_df = df.sort_values(
    by="Income",
    ascending=False
)

print(sorted_df)
```

```python
print("Mean =",
      df["Income"].mean())

print("Median =",
      df["Income"].median())

print("Mode =",
      df["Income"].mode()[0])

print("Standard Deviation =",
      df["Income"].std())
```

```python
print(df.describe())
```

```python
type_avg = df.groupby(
    "Education"
)["Income"].mean()

print(type_avg)
```
